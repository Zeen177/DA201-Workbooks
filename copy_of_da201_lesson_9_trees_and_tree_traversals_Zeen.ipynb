### 1.1 — Tree versus list
- Structurally, what makes a tree different from a linked list? Answer in terms of pointers per node.
- What does that difference let a tree do that a list cannot?


**Your answer — 1.1 — Tree versus list**

A list node has only 1 pointer to the next node. A tree node has multiple pointers to child nodes.

Represent Hierarchy: It can model parent-child relationships.
Faster Search: It can skip whole branches during search ($O(\log n)$ time), while a list must search step-by-step ($O(n)$).


### 1.2 — Height and depth
- Define a tree's *height* and a node's *depth*.
- For a tree with one node, give both values and explain your convention.


**Your answer — 1.2 — Height and depth**

Tree Height: The number of edges on the longest path from the root to a leaf node.

Node Depth: The number of edges on the path from the root to that specific node.

For a single-node tree:

Height = 0

Depth = 0

Convention: We measure depth and height by counting edges, not nodes. A single node has no edges connected to it, so both values are 0. (An empty tree would be -1).


### 1.3 — Why order matters
- For an array, reading order is essentially fixed. For a tree it is a real choice. Explain why.
- Give one task where inorder is the only sensible choice, and one where postorder is.


**Your answer — 1.3 — Why order matters**

Arrays are 1D lines, so there is only one way to read them. Trees have branches, so we can choose to process a node before, between, or after its children.

Tasks & Why:
Inorder Task: Getting sorted numbers from a Binary Search Tree (BST).
Why: A BST follows "Left < Root < Right", so reading in this order automatically gives ascending values.

Postorder Task: Deleting or freeing a tree's memory.
Why: Children must be deleted first. If you delete the root first, you lose the pointers to its children.

## Section 3 — Core Implementation  (20 points)

Implement the three recursive traversals and the level-order traversal in `tree.cpp`. The example tree is already built for you.


````python
%%writefile tree.h
#ifndef TREE_H
#define TREE_H
#include <iostream>

class Tree {
private:
    struct Node {
        char data;
        Node* left;
        Node* right;
        Node(char d) : data(d), left(nullptr), right(nullptr) {}
    };
    Node* root;
    void clear(Node* node);
    void preorder(Node* node) const;
    void inorder(Node* node) const;
    void postorder(Node* node) const;
public:
    Tree();
    ~Tree();
    void buildExampleTree();
    void printPreorder() const;
    void printInorder() const;
    void printPostorder() const;
    void printLevelOrder() const;
};
#endif
````

````python
%%writefile tree.cpp
#include "tree.h"
#include <queue>
using namespace std;

Tree::Tree() : root(nullptr) {}
Tree::~Tree() { clear(root); }

void Tree::clear(Node* node) {
    if (node == nullptr) return;
    clear(node->left);
    clear(node->right);
    delete node;
}

void Tree::buildExampleTree() {
    root = new Node('*');
    root->left = new Node('+');
    root->right = new Node('e');
    root->left->left = new Node('a');
    root->left->right = new Node('-');
    root->left->right->left = new Node('b');
    root->left->right->right = new Node('c');
}

// TODO: visit root, then left, then right
void Tree::preorder(Node* node) const {
    if (node == nullptr) return;
    std::cout<< node->data << " ";
    preorder(node->left);
    preorder(node->right);
}

// TODO: visit left, then root, then right
void Tree::inorder(Node* node) const {
    if (node == nullptr) return;
    inorder(node->left);
    std::cout<< node->data << " ";
    inorder(node->right);
}

// TODO: visit left, then right, then root
void Tree::postorder(Node* node) const {
    if (node == nullptr) return;
    postorder(node->left);
    postorder(node->right);
    std::cout << node->data << " ";

}

void Tree::printPreorder()  const { preorder(root); }
void Tree::printInorder()   const { inorder(root); }
void Tree::printPostorder() const { postorder(root); }

// TODO: breadth-first, using the queue included above
void Tree::printLevelOrder() const {
    if ( root == nullptr) return;
    queue<Node*> q;
    q.push(root);
    while (!q.empty()) {
        Node* current = q.front();
        q.pop();
        std::cout << current->data << " ";

        if (current->left != nullptr) {
            q.push(current->left);
        }
        if (current->right != nullptr) {
            q.push(current->right);
        }
    }
}
````

````python
%%writefile main.cpp
#include <iostream>
#include "tree.h"
using namespace std;

int main() {
    Tree t;
    t.buildExampleTree();
    cout << "Preorder:    "; t.printPreorder();    cout << endl;
    cout << "Inorder:     "; t.printInorder();     cout << endl;
    cout << "Postorder:   "; t.printPostorder();   cout << endl;
    cout << "Level-order: "; t.printLevelOrder();  cout << endl;
    return 0;
}
````
